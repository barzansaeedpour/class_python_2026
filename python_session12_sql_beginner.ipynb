## 1. What is SQLite?
SQLite is a lightweight database built into Python. No installation needed.

<div dir="rtl">
SQLite یک پایگاه داده سبک است که داخل خود پایتون وجود دارد و نیازی به نصب جداگانه ندارد.
</div>

```python
import sqlite3

# Create (or connect to) a database file
conn = sqlite3.connect('example.db')

# Create a cursor to execute SQL commands
cursor = conn.cursor()

print("Database connected!")
```

## 2. Creating a Table

<div dir="rtl">
حالا یک جدول برای ذخیره اطلاعات کاربران می‌سازیم.
</div>

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER
)
''')

conn.commit()
print("Table created!")
```

## 3. Inserting Data

<div dir="rtl">
در این بخش اطلاعات جدید داخل جدول قرار می‌دهیم.
</div>

```python
cursor.execute("INSERT INTO users (name, age) VALUES (?, ?)", ("Amir", 25))
cursor.execute("INSERT INTO users (name, age) VALUES (?, ?)", ("Sara", 30))

conn.commit()
print("Data inserted!")
```

## 4. Reading Data

<div dir="rtl">
حالا اطلاعات ذخیره شده را می‌خوانیم.
</div>

```python
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

## 5. Updating Data

<div dir="rtl">
می‌توانیم اطلاعات یک کاربر را تغییر دهیم.
</div>

```python
cursor.execute("UPDATE users SET age = ? WHERE name = ?", (26, "Alice"))
conn.commit()
print("Data updated!")
```

## 6. Deleting Data

<div dir="rtl">
حذف یک رکورد از جدول.
</div>

```python
cursor.execute("DELETE FROM users WHERE name = ?", ("Bob",))
conn.commit()
print("Data deleted!")
```

## 7. Closing the Connection

<div dir="rtl">
در پایان باید اتصال به دیتابیس را ببندیم.
</div>

```python
conn.close()
print("Connection closed!")
```
